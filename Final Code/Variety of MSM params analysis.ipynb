Shift in green one likely due to the setting of higher transition probabilties so multipliers change more frequently leading to rougher paths and hence a lower average Holder exponent. Reduced tail on right for red is likely due to higher 'b' value in MSM causing long term multipliers to change even less. This results in fewer small magnitude volatility shifts (which are caused by the occasional change in these long term multipliers).
