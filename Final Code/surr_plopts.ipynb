```python
import matplotlib.pyplot as plt
from iaaft_copy import surrogates
import numpy as np
```

```python
x_1_4 = np.sqrt(np.load("all_paths_msm_1_4.npz")['h']) * 100

_, og_1_4, shuffled_1_4, ps_1_4, rescaled_1_4 = surrogates(x_1_4[:1000, 1], ns=1, tol_pc=5., verbose=True)
```

```python
plt.style.use('seaborn-v0_8-dark')

fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

axes[0].plot(og_1_4, color='blue', alpha=0.7)
axes[0].set_title("Original", fontsize=18)

axes[1].plot(shuffled_1_4, color='green', alpha=0.7)
axes[1].set_title("Shuffled", fontsize=18)

axes[2].plot(np.real(ps_1_4), color='red', alpha=0.7)
axes[2].set_title("Enforce Power Spectrum", fontsize=18)

axes[3].plot(np.real(rescaled_1_4), color='black', alpha=0.7)
axes[3].set_title("Enforce Distribution", fontsize=18)

for ax in axes:
    ax.grid(True, alpha=0.7)
    ax.tick_params(axis='both', which='major', labelsize=12)
    ax.set_ylabel("Volatility (%)", fontsize=14)

plt.xlabel("Time (days)", fontsize=14)
plt.suptitle("One Loop of the IAAFT Algorithm for a Daily Volatility Series", fontsize=22, fontweight="bold")
plt.tight_layout()
plt.show()
```
